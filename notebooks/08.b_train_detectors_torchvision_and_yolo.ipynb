# 08.b — Train detectors (torchvision + YOLO)

Detector training commands (prints --help first so flags remain accurate).


```python
from pathlib import Path
import subprocess

# --- Robust REPO_ROOT resolution ---
if (Path.cwd() / "src").exists():
    REPO_ROOT = Path.cwd()
elif (Path.cwd().parent / "src").exists():
    REPO_ROOT = Path.cwd().parent
else:
    raise RuntimeError("Could not locate repo root.")

print("REPO_ROOT:", REPO_ROOT)


def sh(cmd: str, check: bool = True) -> None:
    """Run a shell command (prints it first) from REPO_ROOT."""
    print("\n▶", cmd)
    subprocess.run(cmd, shell=True, check=check, cwd=REPO_ROOT)


# --- Verify training scripts (run as modules, not raw scripts) ---
for s in [
    "src/torchvision_det/train_frcnn_resnet50.py",
    "src/torchvision_det/train_frcnn_mobilenet.py",
    "src/torchvision_det/train_ssd.py",
]:
    p = REPO_ROOT / s
    if p.exists():
        module = s.replace("/", ".").replace(".py", "")
        sh(f'python -m {module} --help', check=False)
    else:
        print("Missing:", p)


# --- Weight check ---
print("\nChecking for trained weights...")

def newest_path(glob_pat: str):
    paths = list(REPO_ROOT.glob(glob_pat))
    if not paths:
        return None
    paths.sort(key=lambda p: p.stat().st_mtime, reverse=True)
    return paths[0]

best_weight = newest_path("runs/**/best*.pt")
last_weight = newest_path("runs/**/last*.pt")

if best_weight:
    print("Found best weight:", best_weight)

if last_weight:
    print("Found last weight:", last_weight)

if not best_weight and not last_weight:
    print("⚠ No trained weights found.")
```
