```python
from google.colab import drive
drive.mount('/content/drive')

import os, glob, shutil, hashlib, pathlib

src_dir = "/content/REDNET-ML/runs/detect/frcnn_resnet50"
dst_dir = "/content/drive/MyDrive/REDNET-ML-ckpts/frcnn_resnet50"  # <- where to keep results on Drive
os.makedirs(dst_dir, exist_ok=True)

# Pick the newest .pt file (your best/latest model) and copy it
pt_files = glob.glob(src_dir + "/*.pt")
assert pt_files, "No .pt files found — did the training finish and save?"
newest_pt = max(pt_files, key=os.path.getmtime)
shutil.copy2(newest_pt, dst_dir)

# (Optional) also copy common artifacts (plots/csv) from the run
for pattern in ("*.png","*.csv","*.txt"):
    for p in glob.glob(os.path.join(src_dir, pattern)):
        shutil.copy2(p, dst_dir)

print("✅ Copied:")
print("  ", newest_pt, "->", os.path.join(dst_dir, os.path.basename(newest_pt)))

# Verify integrity with MD5
def md5(path):
    h = hashlib.md5()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1<<20), b""):
            h.update(chunk)
    return h.hexdigest()

dst_pt = os.path.join(dst_dir, os.path.basename(newest_pt))
print("MD5 (src):", md5(newest_pt))
print("MD5 (dst):", md5(dst_pt))
```

```python
# Save the WHOLE run to a unique folder on Drive and verify
from google.colab import drive
drive.mount('/content/drive')

import os, time, glob, shutil, hashlib, json, pathlib

SRC = "/content/REDNET-ML/runs/detect/frcnn_resnet50"
RUN_ID = time.strftime("%Y%m%d-%H%M%S")
DST_BASE = "/content/drive/MyDrive/REDNET-ML-ckpts/frcnn_resnet50"
DST = f"{DST_BASE}/{RUN_ID}"
os.makedirs(DST, exist_ok=True)

# Copy all common artifacts
patterns = ("*.pt","*.png","*.csv","*.txt")
copied = []
for pat in patterns:
    for p in glob.glob(os.path.join(SRC, pat)):
        target = os.path.join(DST, os.path.basename(p))
        shutil.copy2(p, target)
        copied.append(target)

# Pick newest .pt and compute MD5s for sanity
def md5(path):
    h = hashlib.md5()
    with open(path, "rb") as f:
        for c in iter(lambda: f.read(1<<20), b""): h.update(c)
    return h.hexdigest()

pt_files = sorted(glob.glob(os.path.join(DST, "*.pt")), key=os.path.getmtime, reverse=True)
assert pt_files, "No .pt in destination — did training produce checkpoints?"
best_pt = pt_files[0]
manifest = {
    "run_id": RUN_ID,
    "saved_from": SRC,
    "dest": DST,
    "files": [os.path.basename(x) for x in copied],
    "best_pt": os.path.basename(best_pt),
    "best_pt_md5": md5(best_pt),
}
with open(os.path.join(DST, "manifest.json"), "w") as f:
    json.dump(manifest, f, indent=2)

print("✅ Saved run to:", DST)
print("   Best checkpoint:", manifest["best_pt"], "| md5:", manifest["best_pt_md5"])
```
